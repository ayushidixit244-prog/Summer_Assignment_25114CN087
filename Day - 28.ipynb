```python
#Question-109:To create library management system.
def library_management_system():
    library = {}

    def add_book(book_id, title, author):
        if book_id in library:
            print(f"Book ID {book_id} already exists.")
        else:
            library[book_id] = {'title': title, 'author': author}
            print(f"Book '{title}' added successfully.")

    def remove_book(book_id):
        if book_id in library:
            removed_book = library.pop(book_id)
            print(f"Book '{removed_book['title']}' removed successfully.")
        else:
            print(f"Book ID {book_id} not found.")

    def view_books():
        if not library:
            print("No books available in the library.")
        else:
            for book_id, details in library.items():
                print(f"ID: {book_id}, Title: {details['title']}, Author: {details['author']}")

    while True:
        print("\nLibrary Management System")
        print("1. Add Book")
        print("2. Remove Book")
        print("3. View Books")
        print("4. Exit")

        choice = input("Enter your choice (1-4): ")

        if choice == '1':
            book_id = input("Enter Book ID: ")
            title = input("Enter Book Title: ")
            author = input("Enter Book Author: ")
            add_book(book_id, title, author)
        elif choice == '2':
            book_id = input("Enter Book ID to remove: ")
            remove_book(book_id)
        elif choice == '3':
            view_books()
        elif choice == '4':
            print("Exiting the Library Management System.")
            break
        else:
            print("Invalid choice. Please try again.")
if __name__ == "__main__":
    library_management_system()         
```

```python
#Question-110:To create bank account system.
def bank_account_system():
    accounts = {}

    def create_account(account_number, name, initial_deposit=0):
        if account_number in accounts:
            print(f"Account {account_number} already exists.")
        else:
            accounts[account_number] = {'name': name, 'balance': initial_deposit}
            print(f"Account {account_number} created for {name} with balance {initial_deposit}.")

    def deposit(account_number, amount):
        account = accounts.get(account_number)
        if account:
            if amount > 0:
                account['balance'] += amount
                print(f"Deposited {amount}. New balance: {account['balance']}.")
            else:
                print("Deposit amount must be positive.")
        else:
            print(f"Account {account_number} not found.")

    def withdraw(account_number, amount):
        account = accounts.get(account_number)
        if account:
            if amount > 0 and amount <= account['balance']:
                account['balance'] -= amount
                print(f"Withdrew {amount}. New balance: {account['balance']}.")
            elif amount <= 0:
                print("Withdrawal amount must be positive.")
            else:
                print("Insufficient funds.")
        else:
            print(f"Account {account_number} not found.")

    def view_account(account_number):
        account = accounts.get(account_number)
        if account:
            print(f"Account: {account_number}, Name: {account['name']}, Balance: {account['balance']}")
        else:
            print(f"Account {account_number} not found.")

    while True:
        print("\nBank Account System")
        print("1. Create Account")
        print("2. Deposit")
        print("3. Withdraw")
        print("4. View Account")
        print("5. Exit")

        choice = input("Enter your choice (1-5): ")

        if choice == '1':
            account_number = input("Enter account number: ")
            name = input("Enter account holder name: ")
            initial_deposit = float(input("Enter initial deposit amount: "))
            create_account(account_number, name, initial_deposit)
        elif choice == '2':
            account_number = input("Enter account number: ")
            amount = float(input("Enter deposit amount: "))
            deposit(account_number, amount)
        elif choice == '3':
            account_number = input("Enter account number: ")
            amount = float(input("Enter withdrawal amount: "))
            withdraw(account_number, amount)
        elif choice == '4':
            account_number = input("Enter account number: ")
            view_account(account_number)
        elif choice == '5':
            print("Exiting the Bank Account System.")
            break
        else:
            print("Invalid choice. Please try again.")

if __name__ == "__main__":
    bank_account_system()
```

```python
#Question-111:To create ticket booking system.
def ticket_booking_system():
    events = {
        "E1": {"name": "Concert", "price": 50, "available": 100},
        "E2": {"name": "Movie", "price": 12, "available": 80},
        "E3": {"name": "Theater Play", "price": 30, "available": 60},
    }
    bookings = {}

    def view_events():
        print("\nAvailable Events:")
        for event_id, event in events.items():
            print(f"{event_id}: {event['name']} - ${event['price']} - {event['available']} tickets left")

    def book_ticket():
        event_id = input("Enter event ID to book: ")
        event = events.get(event_id)
        if not event:
            print("Event not found.")
            return

        try:
            quantity = int(input("Enter number of tickets: "))
        except ValueError:
            print("Please enter a valid number.")
            return

        if quantity <= 0:
            print("Ticket quantity must be positive.")
        elif quantity > event["available"]:
            print("Not enough tickets available.")
        else:
            event["available"] -= quantity
            bookings[event_id] = bookings.get(event_id, 0) + quantity
            total_cost = event["price"] * quantity
            print(f"Booked {quantity} ticket(s) for {event['name']}. Total cost: ${total_cost}.")

    def cancel_booking():
        event_id = input("Enter event ID to cancel booking: ")
        if event_id not in bookings:
            print("No booking found for that event.")
            return

        try:
            quantity = int(input("Enter number of tickets to cancel: "))
        except ValueError:
            print("Please enter a valid number.")
            return

        if quantity <= 0 or quantity > bookings[event_id]:
            print("Invalid cancellation quantity.")
        else:
            bookings[event_id] -= quantity
            events[event_id]["available"] += quantity
            if bookings[event_id] == 0:
                del bookings[event_id]
            print(f"Cancelled {quantity} ticket(s) for {events[event_id]['name']}.")

    def view_bookings():
        if not bookings:
            print("No bookings yet.")
        else:
            print("\nYour Bookings:")
            for event_id, quantity in bookings.items():
                print(f"{events[event_id]['name']}: {quantity} ticket(s)")

    while True:
        print("\nTicket Booking System")
        print("1. View Events")
        print("2. Book Ticket")
        print("3. Cancel Booking")
        print("4. View Bookings")
        print("5. Exit")

        choice = input("Enter your choice (1-5): ")

        if choice == "1":
            view_events()
        elif choice == "2":
            book_ticket()
        elif choice == "3":
            cancel_booking()
        elif choice == "4":
            view_bookings()
        elif choice == "5":
            print("Exiting the Ticket Booking System.")
            break
        else:
            print("Invalid choice. Please try again.")

ticket_booking_system()
```

```python
#Question-112:To create contact management system.
def contact_management_system():
    contacts = {}

    def add_contact(name, phone, email=""):
        if name in contacts:
            print(f"Contact '{name}' already exists.")
        else:
            contacts[name] = {"phone": phone, "email": email}
            print(f"Contact '{name}' added successfully.")

    def remove_contact(name):
        if name in contacts:
            contacts.pop(name)
            print(f"Contact '{name}' removed successfully.")
        else:
            print(f"Contact '{name}' not found.")

    def update_contact(name):
        contact = contacts.get(name)
        if not contact:
            print(f"Contact '{name}' not found.")
            return

        phone = input("Enter new phone number (leave blank to keep current): ").strip()
        email = input("Enter new email (leave blank to keep current): ").strip()

        if phone:
            contact["phone"] = phone
        if email:
            contact["email"] = email

        print(f"Contact '{name}' updated successfully.")

    def view_contacts():
        if not contacts:
            print("No contacts found.")
        else:
            print("\nContacts:")
            for name, details in contacts.items():
                print(f"Name: {name}, Phone: {details['phone']}, Email: {details['email']}")

    def search_contact(name):
        contact = contacts.get(name)
        if contact:
            print(f"Name: {name}, Phone: {contact['phone']}, Email: {contact['email']}")
        else:
            print(f"Contact '{name}' not found.")

    while True:
        print("\nContact Management System")
        print("1. Add Contact")
        print("2. Remove Contact")
        print("3. Update Contact")
        print("4. View Contacts")
        print("5. Search Contact")
        print("6. Exit")

        choice = input("Enter your choice (1-6): ").strip()

        if choice == "1":
            name = input("Enter contact name: ").strip()
            phone = input("Enter phone number: ").strip()
            email = input("Enter email (optional): ").strip()
            add_contact(name, phone, email)
        elif choice == "2":
            name = input("Enter contact name to remove: ").strip()
            remove_contact(name)
        elif choice == "3":
            name = input("Enter contact name to update: ").strip()
            update_contact(name)
        elif choice == "4":
            view_contacts()
        elif choice == "5":
            name = input("Enter contact name to search: ").strip()
            search_contact(name)
        elif choice == "6":
            print("Exiting the Contact Management System.")
            break
        else:
            print("Invalid choice. Please try again.")

contact_management_system()
```
