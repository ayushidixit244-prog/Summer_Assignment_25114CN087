```python
#Question-37:Print star pyramid.
n = 5
for i in range(n):      
    print(" " * (n - i - 1) + "*" * (2 * i + 1))    
    
```

```python
#Question-38:Print reverse star pyramid.
n = 5
for i in range(n, 0, -1):
    print(" " * (n - i) + "*" * (2 * i - 1))
```

```python
#Question-39:Print number pyramid.
n = 5
for i in range(1, n + 1):       
    print(" " * (n - i) + "".join(str(j) for j in range(1, i + 1)) + "".join(str(j) for j in range(i - 1, 0, -1)))
```

```python
#Question-40:Print character pyramid.
n = 5
for i in range(1, n + 1):       
    print(" " * (n - i) + "".join(chr(64 + j) for j in range(1, i + 1)) + "".join(chr(64 + j) for j in range(i - 1, 0, -1)))
```
