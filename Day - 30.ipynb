```python
#Question-117:To create student record system using arrays and string.
student_ids = []
student_names = []
student_courses = []
student_marks = []

def add_student(student_id, name, course, marks):
    student_ids.append(str(student_id))
    student_names.append(str(name))
    student_courses.append(str(course))
    student_marks.append(str(marks))

def display_students():
    print("ID\tName\tCourse\tMarks")
    for i in range(len(student_ids)):
        print(f"{student_ids[i]}\t{student_names[i]}\t{student_courses[i]}\t{student_marks[i]}")

def find_student_by_id(student_id):
    student_id = str(student_id)
    if student_id in student_ids:
        idx = student_ids.index(student_id)
        return {
            "id": student_ids[idx],
            "name": student_names[idx],
            "course": student_courses[idx],
            "marks": student_marks[idx]
        }
    return None

def update_student_marks(student_id, new_marks):
    student_id = str(student_id)
    if student_id in student_ids:
        idx = student_ids.index(student_id)
        student_marks[idx] = str(new_marks)
        return True
    return False

# sample records
add_student(101, "Alice", "Math", 88)
add_student(102, "Bob", "Physics", 92)
add_student(103, "Charlie", "Chemistry", 79)

display_students()
result = find_student_by_id(102)
if result:
    print("\nFound student:", result)
else:
    print("\nStudent not found.")

update_student_marks(103, 85)
print("\nAfter update:")
display_students()
```

```python
#Question-118:To create mini library system.
book_ids = []
book_titles = []
book_authors = []
book_status = []

def add_book(book_id, title, author, status="available"):
    book_ids.append(str(book_id))
    book_titles.append(str(title))
    book_authors.append(str(author))
    book_status.append(str(status))

def display_books():
    print("ID\tTitle\tAuthor\tStatus")
    for i in range(len(book_ids)):
        print(f"{book_ids[i]}\t{book_titles[i]}\t{book_authors[i]}\t{book_status[i]}")

def find_book_by_id(book_id):
    book_id = str(book_id)
    if book_id in book_ids:
        idx = book_ids.index(book_id)
        return {
            "id": book_ids[idx],
            "title": book_titles[idx],
            "author": book_authors[idx],
            "status": book_status[idx]
        }
    return None

def update_book_status(book_id, status):
    book_id = str(book_id)
    if book_id in book_ids:
        idx = book_ids.index(book_id)
        book_status[idx] = str(status)
        return True
    return False

def remove_book(book_id):
    book_id = str(book_id)
    if book_id in book_ids:
        idx = book_ids.index(book_id)
        del book_ids[idx]
        del book_titles[idx]
        del book_authors[idx]
        del book_status[idx]
        return True
    return False

# sample records
add_book(201, "Pride and Prejudice", "Jane Austen")
add_book(202, "1984", "George Orwell")
add_book(203, "The Hobbit", "J.R.R. Tolkien", "checked out")

display_books()

result = find_book_by_id(202)
if result:
    print("\nFound book:", result)
else:
    print("\nBook not found.")

update_book_status(203, "available")
remove_book(201)

print("\nAfter updates:")
display_books()
```

```python
# Question-120:To develop complete mini project-integrated Management System.

# IN-MEMORY DATA STORAGE (GLOBAL DICTIONARIES)
students_db = {}   # Schema: {sid: {"name": str, "course": str, "marks": float}}
library_db = {}    # Schema: {bid: {"title": str, "author": str, "status": str}}
employees_db = {}  # Schema: {eid: {"name": str, "position": str, "salary": float}}

# STUDENT MANAGEMENT FUNCTIONS
def add_student(sid, name, course, marks):
    try:
        marks = float(marks)
    except ValueError:
        marks = 0.0  # Default fallback if invalid input is given
    students_db[sid] = {"name": name, "course": course, "marks": marks}

def display_students():
    if not students_db:
        print("\nNo student records found.")
        return
    print(f"\n{'ID':<10} {'Name':<20} {'Course':<15} {'Marks':<10}")
    print("-" * 55)
    for sid, info in students_db.items():
        print(f"{sid:<10} {info['name']:<20} {info['course']:<15} {info['marks']:<10.2f}")

def find_student_by_id(sid):                
    if sid in students_db:
        info = students_db[sid]
        return f"ID: {sid} | Name: {info['name']} | Course: {info['course']} | Marks: {info['marks']:.2f}"
    return None

def update_student_marks(sid, marks):
    if sid in students_db:
        try:
            students_db[sid]["marks"] = float(marks)
            return True
        except ValueError:
            print("Invalid marks format. Update failed.")
            return False
    return False

# LIBRARY MANAGEMENT FUNCTIONS
def add_book(bid, title, author):
    library_db[bid] = {"title": title, "author": author, "status": "available"}

def display_books():
    if not library_db:
        print("\nNo books found in the library.")
        return
    print(f"\n{'Book ID':<10} {'Title':<25} {'Author':<20} {'Status':<15}")
    print("-" * 70)
    for bid, info in library_db.items():
        print(f"{bid:<10} {info['title']:<25} {info['author']:<20} {info['status']:<15}")

def find_book_by_id(bid):
    if bid in library_db:
        info = library_db[bid]
        return f"ID: {bid} | Title: {info['title']} | Author: {info['author']} | Status: {info['status']}"
    return None

def update_book_status(bid, status):
    if bid in library_db:
        library_db[bid]["status"] = status
        return True
    return False

def remove_book(bid):
    if bid in library_db:
        del library_db[bid]
        return True
    return False

# EMPLOYEE MANAGEMENT FUNCTIONS
def add_employee(eid, name, position, salary):
    try:
        salary = float(salary)
    except ValueError:
        salary = 0.0
    employees_db[eid] = {"name": name, "position": position, "salary": salary}

def display_employees():
    if not employees_db:
        print("\nNo employee records found.")
        return
    print(f"\n{'Emp ID':<10} {'Name':<20} {'Position':<20} {'Salary':<10}")
    print("-" * 65)
    for eid, info in employees_db.items():
        print(f"{eid:<10} {info['name']:<20} {info['position']:<20} ${info['salary']:<10.2f}")

def find_employee_by_id(eid):
    if eid in employees_db:
        info = employees_db[eid]
        return f"ID: {eid} | Name: {info['name']} | Position: {info['position']} | Salary: ${info['salary']:.2f}"
    return None

def update_employee_salary(eid, salary):
    if eid in employees_db:
        try:
            employees_db[eid]["salary"] = float(salary)
            return True
        except ValueError:
            print("Invalid salary format. Update failed.")
            return False
    return False

def remove_employee(eid):
    if eid in employees_db:
        del employees_db[eid]
        return True
    return False
# MENU AND NAVIGATION CONTROLLERS

def display_menu():
    print("\n" + "="*50)
    print("INTEGRATED MANAGEMENT SYSTEM")
    print("="*50)
    print("1. Student Records")
    print("2. Library System")
    print("3. Employee Management")
    print("4. Exit")
    print("="*50)

def student_menu():
    print("\nSTUDENT MENU:")
    print("1. Add Student")
    print("2. Display All Students")
    print("3. Find Student by ID")
    print("4. Update Student Marks")
    print("5. Back to Main Menu")

def library_menu():
    print("\nLIBRARY MENU:")
    print("1. Add Book")
    print("2. Display All Books")
    print("3. Find Book by ID")
    print("4. Update Book Status")
    print("5. Remove Book")
    print("6. Back to Main Menu")

def employee_menu():
    print("\nEMPLOYEE MENU:")
    print("1. Add Employee")
    print("2. Display All Employees")
    print("3. Find Employee by ID")
    print("4. Update Employee Salary")
    print("5. Remove Employee")
    print("6. Back to Main Menu")

def run_system():
    while True:
        display_menu()
        choice = input("Enter your choice (1-4): ").strip()
        
        if choice == "1":
            while True:
                student_menu()
                sub_choice = input("Enter your choice (1-5): ").strip()
                if sub_choice == "1":
                    sid = input("Enter Student ID: ")
                    name = input("Enter Name: ")
                    course = input("Enter Course: ")
                    marks = input("Enter Marks: ")
                    add_student(sid, name, course, marks)
                    print("Student added successfully!")
                elif sub_choice == "2":
                    display_students()
                elif sub_choice == "3":
                    sid = input("Enter Student ID to find: ")
                    result = find_student_by_id(sid)
                    print(result if result else "Student not found!")
                elif sub_choice == "4":
                    sid = input("Enter Student ID: ")
                    marks = input("Enter New Marks: ")
                    if update_student_marks(sid, marks):
                        print("Marks updated successfully!")
                    else:
                        print("Student not found!")
                elif sub_choice == "5":
                    break
                else:
                    print("Invalid choice! Please try again.")

        elif choice == "2":
            while True:
                library_menu()
                sub_choice = input("Enter your choice (1-6): ").strip()
                if sub_choice == "1":
                    bid = input("Enter Book ID: ")
                    title = input("Enter Title: ")
                    author = input("Enter Author: ")
                    add_book(bid, title, author)
                    print("Book added successfully!")
                elif sub_choice == "2":
                    display_books()
                elif sub_choice == "3":
                    bid = input("Enter Book ID to find: ")
                    result = find_book_by_id(bid)
                    print(result if result else "Book not found!")
                elif sub_choice == "4":
                    bid = input("Enter Book ID: ")
                    status = input("Enter Status (available/checked out): ")
                    if update_book_status(bid, status):
                        print("Status updated successfully!")
                    else:
                        print("Book not found!")
                elif sub_choice == "5":
                    bid = input("Enter Book ID to remove: ")
                    if remove_book(bid):
                        print("Book removed successfully!")
                    else:
                        print("Book not found!")
                elif sub_choice == "6":
                    break
                else:
                    print("Invalid choice! Please try again.")

        elif choice == "3":
            while True:
                employee_menu()
                sub_choice = input("Enter your choice (1-6): ").strip()
                if sub_choice == "1":
                    eid = input("Enter Employee ID: ")
                    name = input("Enter Name: ")
                    position = input("Enter Position: ")
                    salary = input("Enter Salary: ")
                    add_employee(eid, name, position, salary)
                    print("Employee added successfully!")
                elif sub_choice == "2":
                    display_employees()
                elif sub_choice == "3":
                    eid = input("Enter Employee ID to find: ")
                    result = find_employee_by_id(eid)
                    print(result if result else "Employee not found!")
                elif sub_choice == "4":
                    eid = input("Enter Employee ID: ")
                    salary = input("Enter New Salary: ")
                    if update_employee_salary(eid, salary):
                        print("Salary updated successfully!")
                    else:
                        print("Employee not found!")
                elif sub_choice == "5":
                    eid = input("Enter Employee ID to remove: ")
                    if remove_employee(eid):
                        print("Employee removed successfully!")
                    else:
                        print("Employee not found!")
                elif sub_choice == "6":
                    break
                else:
                    print("Invalid choice! Please try again.")

        elif choice == "4":
            print("Exiting system. Goodbye!")
            break
        else:
            print("Invalid choice! Please try again.")

if __name__ == "__main__":                      # Run the system
    run_system()
```
