```python
#Question-97:To merge two sorted arrays.
def merge_sorted_arrays(arr1, arr2):
    merged_array = []
    i = j = 0
    while i < len(arr1) and j < len(arr2):
        if arr1[i] < arr2[j]:
            merged_array.append(arr1[i])
            i += 1
        else:
            merged_array.append(arr2[j])
            j += 1
    while i < len(arr1):                    # Append remaining elements from arr1, if any
        merged_array.append(arr1[i])
        i += 1   
    while j < len(arr2):                    # Append remaining elements from arr2, if any   
        merged_array.append(arr2[j])
        j += 1
    return merged_array
arr1 = [1, 3, 5, 7]
arr2 = [2, 4, 6, 8]
merged_array = merge_sorted_arrays(arr1, arr2)
print("Merged sorted array:", merged_array)
```

```python
#Question-98:To find common characters in two strings.
str1 = "hello"
str2 = "world"
common_chars = []
for char in str1:
    if char in str2 and char not in common_chars:
        common_chars.append(char)
print("Common characters:", common_chars)
```

```python
#Question-99:To sort names alphabetically in string format.
names = "Charlie, Alice, Bob, David"
names_list = names.split(", ")
names_list.sort()
sorted_names = ", ".join(names_list)    # Join the sorted list back into a string
print("Sorted names:", sorted_names)
```

```python
#Question-100:To sort words by length in string format.
words = "apple, banana, cherry, date"
words_list = words.split(", ")
words_list.sort(key=len)
sorted_words = ", ".join(words_list)
print("Sorted words by length:", sorted_words)
```
