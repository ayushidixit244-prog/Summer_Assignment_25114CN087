```python
#Question-1:Check perfect number.
n= int(input("Enter number: "))
i,s= 1,0
while i<n:
    if n%i==0:
        s+=i
    i+=1
if s==n:
    print(n,"is perfect number")   
else:
    print(n,"is not perfect number")
```

```python
#Question-2:Check strong number.
import math
n= int(input("Enter number: "))
temp=n
fact =0
while temp>0:
    r= temp%10
    fact+= math.factorial(r)      #Factorial of numbers using math module
    temp//= 10
if fact==n:
    print(n,"is strong number")   
else:
    print(n,"is not strong number")
```

```python
#Question-4:Largest prime factor.
n= int(input("Enter number: "))
a=n
b=2
c=[]                #To store the factors
while a!=1:
    if a%b==0:      #Checking factors
        a= a//b
        c.append(b)
    else:
        b+=1
print(c)
print("Largest prime factor of",n,"is",max(c))            
```
