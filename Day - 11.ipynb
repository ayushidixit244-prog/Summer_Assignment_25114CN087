```python
#Question-41:Function to find sum of two numbers.
def sum(a, b):
    return a + b
n1 = float(input("Enter first number: "))
n2 = float(input("Enter second number: "))
result = sum(n1, n2)
print("The sum of", n1, "and", n2, "is:", result)
```

```python
#Question-42:Function to find maximum.
def maximum(a, b):
    return max(a, b)
n1 = float(input("Enter first number: "))
n2 = float(input("Enter second number: "))
max_value = maximum(n1, n2)
print("The maximum of", n1, "and", n2, "is:", max_value)
```

```python
#Question-43:Function to check prime.
def prime(n):
    if n <= 1:
        return False
    for i in range(2,(n * 1//2 + 1)):             #Applying loop to check if number is divisible by any number from 2 to n/2
        if n % i == 0:
            return False
    return True
num = int(input("Enter a number: "))
if prime(num):
    print(num, "is a prime number.") 
else:
    print(num, "is not a prime number.") 
```

```python
#Question-44:Function to find factorial.
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
num = int(input("Enter a number: "))
result = factorial(num)
print("The factorial of", num, "is:", result)
```
