```python
#Question-1:Print half pyramid pattern.
n = 5
for i in range(1, n + 1):       #Outer loop for rows
    for j in range(1, i + 1):   #Inner loop for columns
        print("*", end="")
    print()
```

```python
#Question-2:Print number triangle.
n = 5
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end="")
    print()
```

```python
#Question-3:Print character triangle.
n = 5
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(chr(64 + j), end="")
    print()
```

```python
#Question-4:Print repeated number triangle.
n = 5
for i in range(1, n + 1):  
    for j in range(1, i + 1):
        print(i, end="")
    print()
```
