```python
#Question-1:Find the sum of digits of a number.
n=int(input("Enter the a number"))
sum=0
num=n
while num>0:
    sum += num%10
    num = num//10
print("The sum of digits of",n ,"is", sum)    
```

```python
#Question-2:To reverse a number.
n=int(input("Enter the a number"))
num =n
rev=0
while num>0:
    rem = num%10
    rev= rev*10 + rem
    num //=10
print("The reverse of ",n ,"is",rev)        
```

```python
#Question-3:Find the product of digits.
n=int(input("Enter the a number"))
num =n
pro=1
while num>0:
    rem = num%10
    pro *= rem
    num //=10
print ("The product of ",n ,"is",pro)      
```

```python
#Question-4:Check whether the number is palindrome or not.
n=int(input("Enter the a number"))
num =n
rev=0
while num>0:
    rem= num%10
    rev= rev*10 + rem
    num //=10
if n==rev:
    print (n ,"is palindrome")   
else:
    print (n ,"is not palindrome")         

```
