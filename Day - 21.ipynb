```python
#Question-81:Find string length without using strlen().
s = input("Enter a string: ")
length = 0
for char in s:
    length += 1
print("Length of the string:", length)
```

```python
#Question-82:Reverse a string.
s = input("Enter a string: ")
reverse= ""
for char in s:
    reverse = char + reverse
print("Reversed string:", reverse)
```

```python
#Question-83:Count vowels and consonants in a string.
s = input("Enter a string: ")
vowels = "aeiouAEIOU"
vowel_count = 0
consonant_count = 0
for char in s:
    if char.isalpha():  # Check if the character is an alphabet
        if char in vowels:
            vowel_count += 1
        else:
            consonant_count += 1
print("Number of vowels:", vowel_count)
print("Number of consonants:", consonant_count)
```

```python
#Question-84:Convert lowercase to uppercase.
s = input("Enter a string: ")
uppercase= ""
for char in s:
    if char.islower():
        uppercase += char.upper()
    else:
        uppercase += char
print("Uppercase string:", uppercase)
```
