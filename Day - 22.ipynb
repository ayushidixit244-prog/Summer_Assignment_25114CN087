```python
#Question-85:Check if a string is a palindrome.
s = input("Enter a string: ")
reverse= ""
for char in s:
    reverse = char + reverse
if s == reverse:
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")
```

```python
#Question-86:Count words in a sentence.
s = input("Enter a sentence: ")
words = s.split()                               # Split the sentence into words using whitespace as the delimiter
print("Number of words in the sentence:", len(words))
```

```python
#Question-87:Find character frequency in a string.
s = input("Enter a string: ")
char = input("Enter a character to find its frequency: ")
count = 0
for c in s:
    if c == char:
        count += 1
print("Frequency of", char, "in the string:", count)
```
