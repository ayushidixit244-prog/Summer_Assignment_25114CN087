```python
#Question-49:Take input and display array.
n = int(input("Enter the number of elements: "))
arr = []
print("Enter the elements:")
for i in range(n):
    element = int(input())
    arr.append(element)
print("Array elements are:")
for i in range(n):
    print(arr[i], end=" ")
    
```

```python
#Question-50:Find sum and average of array.
n = int(input("Enter the number of elements: "))
arr = []    
print("Enter the elements:")
for i in range(n):
    element = int(input())
    arr.append(element) 
for i in range(n):
    print(arr[i], end=" ")
total_sum = sum(arr)
average = total_sum / n
print("\nSum of array elements:", total_sum)
print("Average of array elements:", average)
```

```python
#Question-51:Find largest and smallest of array.
n = int(input("Enter the number of elements: "))
arr = []
print("Enter the elements:")
for i in range(n):
    element = int(input())
    arr.append(element) 
for i in range(n):
    print(arr[i], end=" ")
largest = max(arr)
smallest = min(arr) 
print("\nLargest element in the array:", largest)
print("Smallest element in the array:", smallest)
```

```python
#Question-52:Count even and odd elements in an array.
n = int(input("Enter the number of elements: "))
arr = []    
print("Enter the elements:")
for i in range(n):
    element = int(input())
    arr.append(element)
for i in range(n):
    print(arr[i], end=" ")
even_count = 0  
odd_count = 0
for num in arr:
    if num % 2 == 0:
        even_count += 1
    else:
        odd_count += 1
print("\nNumber of even elements in the array:", even_count)
print("Number of odd elements in the array:", odd_count)
```
