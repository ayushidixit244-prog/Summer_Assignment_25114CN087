```python
#Question-33:Print reverse star pattern.
n = 5
for i in range(n, 0, -1):
    print("*" * i) 
```

```python
#Question-34:Print reverse number triangle.
n = 5
for i in range(n, 0, -1):
    for j in range(1, i + 1):
        print(j, end="")
    print()
```

```python
#Question-35:Print repeated character pattern.
n = 5
for i in range(1, n+1):
    for j in range(1, i + 1):
        print(chr(64 + i), end="")
    print()
```

```python
#Question-36:Print hollow square pattern.
n = 5
for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            print("*", end="")
        else:
            print(" ", end="")
    print()
```
