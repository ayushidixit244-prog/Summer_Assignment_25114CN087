```python
#Question-69:Bubble sorting in array.
n = int(input("Enter the number of elements in the array: "))
arr = []
print("Enter elements of the array:")
for i in range(n):                                  #Take input for array elements
    element = int(input())
    arr.append(element) 
for i in range(n):                                  #Print array before rotation
    print(arr[i], end=" ") 
for i in range(n):
    for j in range(0, n-i-1):
        if arr[j] > arr[j+1]:
            arr[j], arr[j+1] = arr[j+1], arr[j]
print("\nArray after bubble sort:", arr)
```

```python
#Question-70:Selection sorting in array.
n = int(input("Enter the number of elements in the array: "))
arr = []    
print("Enter elements of the array:")
for i in range(n):                                  #Take input for array elements
    element = int(input())
    arr.append(element)
for i in range(n):                                  #Print array before rotation
    print(arr[i], end=" ")
for i in range(n):
    min_index = i
    for j in range(i+1, n):
        if arr[j] < arr[min_index]:                 #Find the minimum element in remaining unsorted array
            min_index = j
    arr[i], arr[min_index] = arr[min_index], arr[i]     #Swap the found minimum element with the first element
print("\nArray after selection sort:", arr)
```

```python
#Question-71:Binary search in array.
n = int(input("Enter the number of elements in the array: "))
arr = []
print("Enter elements of the array:")
for i in range(n):                                  #Take input for array elements
    element = int(input())
    arr.append(element)
for i in range(n):                                  #Print array before rotation
    print(arr[i], end=" ")                                           
arr2 = sorted(arr)  
print("\nArray after sorting:")                                     #Sort the array before performing binary search
for i in range(n):
    print(arr2[i], end=" ")
x= int(input("\nEnter the element to search: "))
low = 0
high = n - 1
found = False
while low <= high:
    mid = (low + high) // 2
    if arr2[mid] == x:
        found = True
        break
    elif arr2[mid] < x:
        low = mid + 1
    else:
        high = mid - 1
if found:
    print(f"\nElement {x} found in the array at index {mid}.")
else:
    print(f"\nElement {x} not found in the array.")
```

```python
#Question-72:Sort array in descending order.
n = int(input("Enter the number of elements in the array: "))
arr = []
print("Enter elements of the array:")
for i in range(n):                                  #Take input for array elements
    element = int(input())
    arr.append(element) 
for i in range(n):                                  #Print array before rotation
    print(arr[i], end=" ") 
arr.sort(reverse = True)

print("\nArray in descending order:", arr)
```
