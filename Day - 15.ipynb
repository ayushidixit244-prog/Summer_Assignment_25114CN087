```python
#Question-57:Reverse array.
n = int(input("Enter the number of elements in the array: "))
arr = []
print("Enter elements of the array:")
for i in range(n):                                  #Take input for array elements
    element = int(input())
    arr.append(element) 
for i in range(n):                                  #Print array
    print(arr[i], end=" ")
print("\nReverse array:")    
for i in range(n-1, -1, -1):                        #Print array in reverse order
    print(arr[i], end=" ")
```

```python
#Question-58:To rotate array left.
n = int(input("Enter the number of elements in the array: "))
arr = []    
print("Enter elements of the array:")
for i in range(n):                                  #Take input for array elements
    element = int(input())
    arr.append(element)
for i in range(n):                                  #Print array before rotation
    print(arr[i], end=" ")    
d = int(input("Enter the number of positions to rotate left: "))
d = d % n   
rotated_arr = arr[d:] + arr[:d]
print("\nRotated array:")                           #Print array after rotation
for i in range(n):
    print(rotated_arr[i], end=" ")           
```

```python
#Question-59:To rotate array right.
n = int(input("Enter the number of elements in the array: "))
arr = []
print("Enter elements of the array:")
for i in range(n):                                  #Take input for array elements
    element = int(input())
    arr.append(element) 
for i in range(n):                                  #Print array before rotation
    print(arr[i], end=" ")  
d = int(input("Enter the number of positions to rotate right: "))
d = d % n
rotated_arr = arr[-d:] + arr[:-d]
print("\nRotated array:")
for i in range(n):                                  #Print array after rotation
    print(rotated_arr[i], end=" ")
```

```python
#Question-60:Move zeros to the end of the array.
n = int(input("Enter the number of elements in the array: "))
arr = []
print("Enter elements of the array:")
for i in range(n):                                  #Take input for array elements
    element = int(input())
    arr.append(element)
print("Array before moving zeros:")
for i in range(n):                                  #Print array before moving zeros
    print(arr[i], end=" ")
non_zero_arr = [x for x in arr if x != 0]           #Create a new array with non-zero elements
zero_count = arr.count(0)                           #Count the number of zeros in the original array
result_arr = non_zero_arr + [0] * zero_count        #Combine non-zero elements with zeros at the end
print("\nArray after moving zeros to the end:")
for i in range(n):                                  #Print array after moving zeros
    print(result_arr[i], end=" ")
            
```
