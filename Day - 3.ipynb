```python
#Question-2:Print prime numbers in the range.
n= int(input("Enter the range"))
print("The prime numbers upto", n)
for i in range(2,n+1):
    for j in range(2,i):
        if i%j==0:
            break
    else:
        print(i)      
```

```python
#Question-3:Find GCD of two numbers.
n1= int(input("Enter the number"))
n2= int(input("Enter the number"))
if n1> n2:
    small=n2
else:
    small=n1
for i in range(1,small+1):
    if ((n1%i==0) and (n2%i==0)):
        hcf=i
print("The GCD or HCF of",n1,"and", n2, "is", hcf)
```

```python
#Question-4:Find LCM of two numbers.
n1= int(input("Enter the number"))
n2= int(input("Enter the number"))
if n1> n2:
    greater=n1
else:
    greater=n2
while(True):
    if ((greater%n1==0) and (greater%n2==0)):
        lcm= greater
        break
    greater +=1
print("The LCM of",n1,"and", n2, "is", lcm)

```
