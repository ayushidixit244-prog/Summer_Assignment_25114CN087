```python
#Question-1:Generate Fibonacci series.
nterms = int(input("Enter the terms "))
# first two terms
n1, n2 = 0, 1
count = 0
if nterms <= 0:
   print("Please enter a positive integer")
elif nterms == 1:
   print("Fibonacci sequence upto",nterms,":")
   print(n1)
else:
   print("Fibonacci sequence:")
   while count < nterms:
       print(n1)
       nth = n1 + n2
       n1 = n2
       n2 = nth
       count += 1
```

```python
#Question-3:Check armstrong number.
n = int(input("Enter a number: "))
sum = 0
l=len(str(n))
temp = n
while temp > 0:
   r = temp % 10
   sum += r ** l
   temp //= 10
if n == sum:
   print(n,"is an Armstrong number")
else:
   print(n,"is not an Armstrong number")
```

```python
#Question-4:Check armstrong number in a range.
upper = int(input("Enter a number"))
lower = int(input("Enter a number"))
print("Armstrong numbers are:")
for i in range(lower,upper+1):
    sum = 0
    l=len(str(i))
    temp = i
    while temp > 0:
       r = temp % 10
       sum += r ** l
       temp //= 10
    if i == sum:
       print(i)
    
   
```
