```python
#Question-61:Find missing number in array.

arr = [1,2,4,5,6,7]                    # Example array with numbers from 1 to 7, missing number is 3

n = len(arr) + 1                       # Since one number is missing
total_sum = n * (n + 1) // 2           # Sum of first n natural numbers
arr_sum = sum(arr)                     # Sum of elements in the array
missing_num = total_sum - arr_sum      # The missing number is the difference
print("Missing number is:", missing_num) 
```

```python
#Question-62:Find maximum frequency element in array.
import statistics
arr = [1, 2, 2, 3, 3, 4, 2, 4, 3, 3, 4, 4, 4]
max_freq_element = statistics.mode(arr)
print("Maximum frequency element is:", max_freq_element)
```

```python
#Question-63:Find pair with given sum in array.
arr = [1, 2, 3, 4, 5]
target_sum = int(input("Enter the target sum: "))
pairs = []
for i in range(len(arr)):
    for j in range(i + 1, len(arr)):
        if arr[i] + arr[j] == target_sum:
            pairs.append((arr[i], arr[j]))
print("Pairs with the given sum are:", pairs)
```
