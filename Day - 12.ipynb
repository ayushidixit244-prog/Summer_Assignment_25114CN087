```python
#Question-45:Function for palindrome.
def palindrome(s):
    rev=0
    while s>0:
        rem= s%10
        rev= rev*10 + rem
        s//=10
    return rev
n=int(input("Enter a number: "))
if n==palindrome(n):
    print(n,"is a palindrome.")        
else:
    print(n,"is not a palindrome.")    
```

```python
#Question-46:Function for armstrong.
def armstrong(n):
    sum=0
    l=len(str(n))
    while n>0:
        rem=n%10
        sum+=rem**l
        n//=10
    return sum

n=int(input("Enter a number: "))
if n==armstrong(n):
    print(n,"is an armstrong number.")  
else: 
    print(n,"is not an armstrong number.")    
    
```

```python
#Question-47:Function for fibonacci.
def fibonacci(n):
    if n<=1:
        return n
    else:
        return fibonacci(n-1)+fibonacci(n-2)
num=int(input("Enter the number of terms: "))
print("Fibonacci sequence:")    
for i in range(num):
    print(fibonacci(i), end=" ")
```

```python
#Question-48:Function for perfect number.
def perfect(n):
    sum=0
    for i in range(1,n):
        if n%i==0:
            sum+=i
    return sum
n=int(input("Enter a number: "))
if n==perfect(n):   
    print(n,"is a perfect number.")
else:
    print(n,"is not a perfect number.")    
```
