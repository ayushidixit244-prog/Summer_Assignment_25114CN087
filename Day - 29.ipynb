```python
#Question-113:To create menu-driven calculator.
def calculator():
    while True:
        print("\nMenu Driven Calculator")
        print("1. Add")
        print("2. Subtract")
        print("3. Multiply")
        print("4. Divide")
        print("5. Exit")
        choice = input("Enter your choice (1-5): ")

        if choice == "5":
            print("Exiting calculator.")
            break

        if choice not in {"1", "2", "3", "4"}:
            print("Invalid choice. Please select a valid option.")
            continue

        try:
            num1 = float(input("Enter first number: "))
            num2 = float(input("Enter second number: "))
        except ValueError:
            print("Invalid input. Please enter numeric values.")
            continue

        if choice == "1":
            result = num1 + num2
            op = "+"
        elif choice == "2":
            result = num1 - num2
            op = "-"
        elif choice == "3":
            result = num1 * num2
            op = "*"
        else:
            if num2 == 0:
                print("Cannot divide by zero.")
                continue
            result = num1 / num2
            op = "/"

        print(f"{num1} {op} {num2} = {result}")

calculator()
```

```python
#Question-114:To create menu-driven array operations system.
def array_operations():
    arr = []
    while True:
        print("\nMenu Driven Array Operation System")
        print("1. Create array")
        print("2. Display array")
        print("3. Insert element")
        print("4. Delete element")
        print("5. Search element")
        print("6. Sum of elements")
        print("7. Average of elements")
        print("8. Exit")
        choice = input("Enter your choice (1-8): ")

        if choice == "8":
            print("Exiting array operation system.")
            break

        if choice == "1":
            values = input("Enter array elements separated by spaces: ").split()
            try:
                arr = [int(x) for x in values]
                print("Array created:", arr)
            except ValueError:
                print("Please enter valid integer values.")
        elif choice == "2":
            print("Array contents:", arr)
        elif choice == "3":
            try:
                value = int(input("Enter element to insert: "))
                position = input("Enter position index (leave blank to append): ")
                if position == "":
                    arr.append(value)
                else:
                    idx = int(position)
                    if 0 <= idx <= len(arr):
                        arr.insert(idx, value)
                    else:
                        print("Invalid index.")
                        continue
                print("Updated array:", arr)
            except ValueError:
                print("Please enter valid integer values.")
        elif choice == "4":
            if not arr:
                print("Array is empty.")
                continue
            try:
                value = int(input("Enter element to delete: "))
                if value in arr:
                    arr.remove(value)
                    print("Updated array:", arr)
                else:
                    print("Element not found.")
            except ValueError:
                print("Please enter a valid integer.")
        elif choice == "5":
            try:
                value = int(input("Enter element to search: "))
                if value in arr:
                    print(f"Element found at index {arr.index(value)}.")
                else:
                    print("Element not found.")
            except ValueError:
                print("Please enter a valid integer.")
        elif choice == "6":
            print("Sum of elements:", sum(arr))
        elif choice == "7":
            if arr:
                print("Average of elements:", sum(arr) / len(arr))
            else:
                print("Array is empty.")
        else:
            print("Invalid choice. Please select a valid option.")

array_operations()
```

```python
#Question-115:To create menu-driven string operations system.
def string_operations():
    s = ""
    while True:
        print("\nMenu Driven String Operation System")
        print("1. Create string")
        print("2. Display string")
        print("3. Concatenate string")
        print("4. Length of string")
        print("5. Find substring")
        print("6. Replace substring")
        print("7. Convert to uppercase")
        print("8. Convert to lowercase")
        print("9. Reverse string")
        print("10. Exit")
        choice = input("Enter your choice (1-10): ")

        if choice == "10":
            print("Exiting string operation system.")
            break

        if choice == "1":
            s = input("Enter a string: ")
            print("String created:", s)
        elif choice == "2":
            print("Current string:", s)
        elif choice == "3":
            addition = input("Enter string to concatenate: ")
            s = s + addition
            print("Updated string:", s)
        elif choice == "4":
            print("Length of string:", len(s))
        elif choice == "5":
            sub = input("Enter substring to find: ")
            idx = s.find(sub)
            if idx != -1:
                print(f"Substring found at index {idx}.")
            else:
                print("Substring not found.")
        elif choice == "6":
            old = input("Enter substring to replace: ")
            new = input("Enter replacement string: ")
            s = s.replace(old, new)
            print("Updated string:", s)
        elif choice == "7":
            s = s.upper()
            print("Uppercase string:", s)
        elif choice == "8":
            s = s.lower()
            print("Lowercase string:", s)
        elif choice == "9":
            s = s[::-1]
            print("Reversed string:", s)
        else:
            print("Invalid choice. Please select a valid option.")

string_operations()
```

```python
#Question-116:To create inventory management system.
def inventory_management_system():
    inventory = {}

    while True:
        print("\nInventory Management System")
        print("1. Add item")
        print("2. Display inventory")
        print("3. Update item")
        print("4. Remove item")
        print("5. Search item")
        print("6. Total inventory value")
        print("7. Exit")
        choice = input("Enter your choice (1-7): ")

        if choice == "7":
            print("Exiting inventory management system.")
            break

        if choice == "1":
            name = input("Enter item name: ").strip()
            if not name:
                print("Item name cannot be empty.")
                continue
            try:
                quantity = int(input("Enter quantity: "))
                price = float(input("Enter unit price: "))
            except ValueError:
                print("Please enter valid numeric values for quantity and price.")
                continue
            inventory[name] = {"quantity": quantity, "price": price}
            print(f"Added item: {name}")

        elif choice == "2":
            if not inventory:
                print("Inventory is empty.")
            else:
                print("Current inventory:")
                for name, info in inventory.items():
                    print(f"- {name}: quantity={info['quantity']}, price={info['price']}")

        elif choice == "3":
            name = input("Enter item name to update: ").strip()
            if name not in inventory:
                print("Item not found.")
                continue
            try:
                quantity = int(input("Enter new quantity (leave blank to keep current): ") or inventory[name]["quantity"])
                price_input = input("Enter new unit price (leave blank to keep current): ")
                price = float(price_input) if price_input else inventory[name]["price"]
            except ValueError:
                print("Please enter valid numeric values.")
                continue
            inventory[name]["quantity"] = quantity
            inventory[name]["price"] = price
            print(f"Updated item: {name}")

        elif choice == "4":
            name = input("Enter item name to remove: ").strip()
            if name in inventory:
                del inventory[name]
                print(f"Removed item: {name}")
            else:
                print("Item not found.")

        elif choice == "5":
            name = input("Enter item name to search: ").strip()
            if name in inventory:
                info = inventory[name]
                print(f"Found item: {name}, quantity={info['quantity']}, price={info['price']}")
            else:
                print("Item not found.")

        elif choice == "6":
            total_value = sum(info["quantity"] * info["price"] for info in inventory.values())
            print(f"Total inventory value: {total_value}")

        else:
            print("Invalid choice. Please select a valid option.")

inventory_management_system()
```
