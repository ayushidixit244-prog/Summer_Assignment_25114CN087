```python
#Question-65:To merge arrays.
import numpy as np
arr1 = np.array([10, 29, 35])
arr2 = np.array([47, 58, 63])
merged_arr = np.concatenate((arr1, arr2))
print("Merged Array:", merged_arr)
```

```python
#Question-66:To union of arrays.
a = np.array([10, 20, 30, 40])
b = np.array([20, 40, 60, 80])
c = np.concatenate((a,b))                   # Joins them as it is.
res = np.unique(c)                          # Remove duplicates and automatically sort the result
print(res)
```

```python
#Question-67:Intersection of arrays.
a = np.array([10, 20, 30, 40])
b = np.array([20, 40, 60, 80])
res = np.intersect1d(a,b)
print(res)
```

```python
#Question-68:Find common elements in array.
a = np.array([10, 21, 37, 45])
b = np.array([21, 44, 69, 80])
common_elements = np.intersect1d(a, b)
print("Common elements:", common_elements)  
```
