```python
#Question-1: Calculate sum of first N natural numbers.
n=int(input("Enter the positive integer(N):"))
if n<1:
    print("Please enter natural number only(n>0)")
else:
    # Applying the formula= (N*(N+1))//2
    sum = (n*(n+1))//2
    print("The sum of the first(n) natural numbers is:", sum)    
```

```python
#Question-2:Print multiplication table of given number.
num =int(input("Enter the number whose table is to be written:"))
for i in range(1,11):
    print(num, 'x', i, '=', num*i)
```

```python
#Question-3:Find the factorial of the number.
num =int(input("Enter the number whose factorial is to be find out"))
fact=1
if num<0:
    print("Factorial does not exist ")
elif num==0:
    print("Factorial of 0 id 1 ")  
else:
    for i in range(1,num+1):
        fact= fact*i
    print("Factorial of",num,"is",fact)         
```

```python
#Question-4: To count digits of a number.
num= int(input("Enter the number"))
count=0
while num!= 0:
    num//= 10
    count += 1
print("Number of digits is", count)    
```
