```python
#Question-1:Convert decimal to binary.
n= int(input("Enter number"))
if n == 0:
      print("0")
binary_digits = []
while n > 0:
        binary_digits.append(str(n % 2))
        n //= 2
print("".join(reversed(binary_digits)))  
# "".join(...)- takes the sequence of strings and concatenates them into one single string with no separator.
```

```python
#Question-2:Convert binary to decimal.
binary= int(input("Enter a binary number: "))
base , decimal= 1, 0
while binary>0:
    rem= binary%10
    decimal += rem * base
    base *= 2
    binary //= 10
print("The decimal equivalent is:", decimal)
```

```python
#Question-3:Count sets bits in a number.
n = int(input("Enter number:"))
#print(n.bit_count()) - direct method to count set bits in a number using built-in function bit_count() in Python 3.10 and above.
count= 0
num= n
while num>0:
    if num%2==1:
        count += 1
    num//=2
print("Number of sets bits in",n,"is:", count)
```

```python
#Question-4:Find x**n without pow() function.
x= int(input("Enter base:"))
n= int(input("Enter exponent:"))
b=n
c=1
while b>0:
    c *= x
    b -= 1
print("Result:", c)
```
