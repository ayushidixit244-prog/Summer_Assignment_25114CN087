```python
#Question-53:Linear search.
def linear_Search(array, n, x):    
    for i in range(0, n):
        if (array[i] == x):
            return i
    return -1

n = int(input("Enter the number of elements: "))
arr = []
for i in range(n):
    element = int(input())
    arr.append(element)
print("Array elements are:")
for i in range(n):                                      # Taking the input of array elements
    print(arr[i], end=" ")
x = int(input("Enter the element to search: "))
result = linear_Search(arr, n, x)                        # Calling the function
if(result == -1):
    print("\nElement not found")
else:
    print("\nElement found at index: ", result)
```

```python
#Question-54:Frequency of elements in an array.
def frequency(arr, n):
    freq = {}
    for i in range(n):
        if arr[i] in freq:
            freq[arr[i]] += 1
        else:
            freq[arr[i]] = 1
    for key, value in freq.items():
        print(f"{key}: {value}")
n = int(input("Enter the number of elements: "))
arr = []
for i in range(n):
    element = int(input())
    arr.append(element)
print("Array elements are:")
for i in range(n):
    print(arr[i], end=" ")    
print("\nFrequency of elements:")
frequency(arr, n)
```

```python
#Question-55:Find second largest element.
n = int(input("Enter the number of elements: "))
arr = []
for i in range(n):
    element = int(input())
    arr.append(element)
print("Array elements are:")
for i in range(n):
    print(arr[i], end=" ") 
first = arr[0]
second = arr[0]
for i in range(n):
    if arr[i] > first:
        second = first
        first = arr[i]
    elif arr[i] > second and arr[i] != first:
            second = arr[i]
print("\nSecond largest element is:", second)
```

```python
#Question-56:Find duplicates in array.
n = int(input("Enter the number of elements: "))
arr = []
for i in range(n):
    element = int(input())
    arr.append(element)
print("Array elements are:")
for i in range(n):
    print(arr[i], end=" ")
print("\nDuplicates in the array are:")
for i in range(n):
    for j in range(i + 1, n):
        if arr[i] == arr[j]:
            print(arr[i], end=" ")
        else:
            continue
```
