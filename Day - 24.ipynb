```python
#Question-93:To check string rotation.
s1 = "abcde"
s2 = "cdeab"
if len(s1) == len(s2) and s2 in (s1 + s1):           #Check if s2 is a substring of s1 concatenated with itself
    print(f"The string '{s2}' is a rotation of '{s1}'.")
else:
    print(f"The string '{s2}' is not a rotation of '{s1}'.")
```

```python
#Question-94:To compress a string.
s = "aaabbbcccaaa"
compressed = ""
current_char = s[0]
count = 1
for char in s[1:]:
    if char == current_char:
        count += 1
    else:
        compressed += f"{current_char}{count}"
        current_char = char
        count = 1
compressed += f"{current_char}{count}"         # Add the last character and its count

print(f"The compressed string is: {compressed}")
```

```python
#Question-95:To find longest word.
s = "The quick brown fox jumps over the lazy dog"
words = s.split()
longest_word = max(words, key=len)
print(f"The longest word is: {longest_word}")
```

```python
#Question-96:To remove duplicate characters from a string.
s = "programming"
unique_chars = ""
for char in s:
    if char not in unique_chars:
        unique_chars += char
print(f"The string with duplicate characters removed is: {unique_chars}")
```
